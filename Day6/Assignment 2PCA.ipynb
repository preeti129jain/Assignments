```python
import numpy as np

#1.	Consider the following 2D data: 
X = np.array([
    [2, 1],
    [3, 2],
    [4, 3],
    [5, 4],
    [6, 5]
])
#Using Python:
#•	Calculate the mean of each feature.
feature_mean = np.mean(X, axis=0)

print("feature_mean :", feature_mean)

#•	Subtract the mean from each feature. 
X_centered = (X - feature_mean)

#•	Display the centered data. 
print("X_centered ::", X_centered)
```

```python
#7.	Given the covariance matrix: 
C = np.array([
    [4, 2],
    [2, 3]
])
#Using NumPy:
#•	Calculate the eigenvalues. 
#•	Calculate the eigenvectors. 
#np.linalg.eig returns:
#   - eigenvalues as a 1D array
#   - eigenvectors as columns in a 2D array
eigenvalues, eigenvectors = np.linalg.eig(C)
# Display results
print("Eigenvalues:", eigenvalues)
print("Eigenvectors (columns correspond to eigenvalues):\n", eigenvectors)


#•	Identify the eigenvector corresponding to the largest eigenvalue. 
# Identify the largest eigenvalue and its corresponding eigenvector
max_index = np.argmax(eigenvalues)  # index of largest eigenvalue
largest_eigenvalue = eigenvalues[max_index]
largest_eigenvector = eigenvectors[:, max_index]  # corresponding column

print("\nLargest Eigenvalue:", largest_eigenvalue)
print("Corresponding Eigenvector:", largest_eigenvector)


```

```python
#13. Use the following data:  
X = np.array([ 
    [2, 1], 
    [3, 2], 
    [4, 3], 
    [5, 4], 
    [6, 5] 
]) 
#Apply: from sklearn.decomposition import PCA 
#Reduce the data from 2 dimensions to 1 dimension. 
#Display: 
from sklearn.decomposition import PCA

#• Original data  
print("Original Data:", X)
#• Transformed data  
pca = PCA(n_components=1)
X_transformed = pca.fit_transform(X)
print("Transformed_data:", X_transformed)

#• Explained variance ratio 
print(pca.explained_variance_ratio_) 
print("The two features are perfectly correlated (x₂ = x₁ - 1).Therefore, all variance lies along a single direction.PCA captures 100% of the variance in the first principal component, so the explained variance ratio is 1.0 (100%).")
```
