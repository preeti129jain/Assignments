```python
#Task 2 — From Vector to Matrix
#Suppose an industrial system has 4 sensors recording measurements for 5 time steps.
#Create the following data:
#Temperature:  [25, 26, 25, 27, 28]
#Pressure:     [101, 102, 101, 103, 104]
#Humidity:     [60, 62, 61, 63, 64]
#Vibration:    [0.2, 0.3, 0.2, 0.4, 0.5]
#Create a tensor with shape:
#(5, 4) where:
#rows    → time
#columns → sensors

Temprature = torch.tensor([25, 26, 25, 27, 28])
Pressure =  torch.tensor([101, 102, 101, 103, 104])
Humidity =  torch.tensor([60, 62, 61, 63, 64])
Vibration = torch.tensor([0.2, 0.3, 0.2, 0.4, 0.5])

my_tensor = torch.stack([Temprature, Pressure, Humidity, Vibration])
my_tensor = my_tensor.T

#Then print:
#•	shape 
print("Shape: ", my_tensor.shape)

#•	number of dimensions 
print("Dimensions: ", my_tensor.shape)

#•	first time step 
print("first time step: ", my_tensor[0])

#•	temperature readings 
print("Temprature reading: ", my_tensor[:,0])

#•	vibration readings 
print("Vibration reading: ", my_tensor[:,3])
#Think: What does each axis represent?
```

```python
#Task 9 — Reshape vs Transpose
#Create:
import numpy as np
X = np.arange(24).reshape(2, 3, 4)
print("X initially::", X)
#Now perform:
Y= X.reshape(3, 2, 4)
print("X after reshape", Y)
Z= X.transpose(1, 0, 2)
print("X after transpose", Z)
#Compare the results.
#Are these two operations doing the same thing?
#No, They are different, one just re arranging the elements in desired shape
#other, is change their position
```

```python
#Task 10 — Broadcasting
#Create:
X = np.ones((3, 4))
#and:
b = np.array([10, 20, 30, 40])
#Calculate: X + b
#Then try: X * b
#Question: Why can a (4,) tensor be combined with a (3,4) tensor?
print("X:", X)
Y= X + b
print("Addition:", Y)
Z= X * b
print ("Multiplication:", Z)
print("Question: Why can a (4,) tensor be combined with a (3,4) tensor?")
print("Answer: (4,) tensor spread to the number of rows in X, So, the addition is possible. Since, inner dimension should meet for multiplication, (3,4) * (4,1) is possible")
```

```python
#Task 12 — Tensor Aggregation
#Create:
X = np.random.randn(100, 20, 5)
#Interpret it as:
#100 machines
#20 time steps
#5 sensors
#Calculate:
#1.	Mean across time. 
mean_time = X.mean(axis=1)
print("mean_time::", mean_time)
#2.	Mean across sensors. 
mean_sensor = X.mean(axis=2)
print("mean_sensor::", mean_sensor)
#3.	Mean across machines. 
mean_machine = X.mean(axis=0)
print("mean_time::", mean_machine)
#4.	Maximum sensor value across time. 
max_time = X.max(axis=1)
print("Max senor across time::", max_time)
#5.	Standard deviation across time. 
std_time = X.std(axis=(0,1))
print("Standard deviation across time::", std_time)
#Predict the resulting shape before running the code.
```

```python
#Task 13 — Normalize a Tensor
#Create sensor data:
X = np.random.randn(100, 20, 5)
#Calculate the mean and standard deviation for each sensor.
#Normalize the tensor using:
#x_normalized=(x-μ)/σ
#Then verify:
#mean ≈ 0

sensor1_mean = X.mean(axis=0)
sensor1_std = X.std(axis=0)
X1_normalized = (X - sensor1_mean) / sensor1_std

# Step 4: Verify normalization
normalized_mean = np.mean(X1_normalized, axis=0)
normalized_std = np.std(X1_normalized, axis=0)


sensor2_mean = X.mean(axis=1)
sensor2_std = X.std(axis=1)

sensor3_mean = X.mean(axis=2)
sensor3_std = X.std(axis=2)

print("Sensor 1 mean ::" , sensor1_mean)
print ("Sensor 1 std deviation", sensor1_std)
print("Sensor 1 normalized_mean ::" , normalized_mean)
print ("Sensor 1 normalized_std", normalized_std)

print("Sensor 2 mean ::" , sensor2_mean)
print ("Sensor 2 std deviation", sensor2_std)

print("Sensor 3 mean ::" , sensor3_mean)
print ("Sensor 3 std deviation", sensor3_std)


#standard deviation ≈ 1
```
